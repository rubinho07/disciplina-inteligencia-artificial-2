```python
from scipy.integrate import quad
import numpy as np

def f(x):
    return x**2 * np.sin(x)

resultado, erro = quad(f, 0, np.pi)

print(resultado)
```

```python
from scipy.integrate import solve_ivp

def funcao(x, y):
    return x - y

resultado = solve_ivp(funcao, [0, 10], [1])

print(resultado.y)
```

```python
from scipy.optimize import root
import numpy as np

def f(x):
    return x**4 - 3*x**3 + 2

resultado = root(f, 1)

print(resultado.x)
```

```python
from scipy.optimize import minimize

def f(x):
    return -(-x**2 + 4*x + 1)

resultado = minimize(f, 0)

print("x:", resultado.x)
print("valor:", -resultado.fun)
```

```python
import numpy as np
from scipy.optimize import curve_fit

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

def funcao(x, a, b, c):
    return a*x**2 + b*x + c

valores, erro = curve_fit(funcao, x, y)

print(valores)
```
